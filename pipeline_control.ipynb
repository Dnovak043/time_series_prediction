# Pipeline control panel

Edit every stage's parameters, save/load YAML configs, launch and monitor runs, and browse results — from JupyterLab or a VS Code notebook over SSH.

- **Stage tabs** (data / featurize / encode / distributions / training): every knob of the pipeline. Hover a field for help; rarely-changed fields are under *Advanced*.
- **run & monitor**: launches `python -m pipeline run|train` as a *detached* subprocess — it keeps running if the kernel dies or SSH drops. Re-open this notebook and *Attach* to any run.
- **results**: plot saved `SEQ_DISTR_*` / `CLS_DISTR_*` distributions.

Everything here is also scriptable: `python -m pipeline --help` (see `PIPELINE_GUIDE.md`).
